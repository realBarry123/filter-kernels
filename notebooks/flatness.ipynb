```python
from filter_kernels import convolvers
from filter_kernels.conv_layer import ConvLayer, virtual_kernel
from filter_kernels.utils import metrics
import matplotlib.pyplot as plt
```

```python
oobleck_encoder = convolvers.load("sao1_oobleck").encoder
conv1 = oobleck_encoder.conv1
res_unit1 = oobleck_encoder.block[0].res_unit1
combined = ConvLayer(conv1) * virtual_kernel(res_unit1, module_type="res_unit") 
flatness = metrics.spectral_flatness(combined.weight).flatten()
plt.hist(flatness, bins=100)
```

```python
hubert_layers = convolvers.load("hubert").feature_extractor.conv_layers
combined = ConvLayer(hubert_layers[0].conv) * ConvLayer(hubert_layers[1].conv)
flatness = metrics.spectral_flatness(combined.weight).flatten()
plt.hist(flatness, bins=100)
```

```python
wav2vec_layers = convolvers.load("wav2vec2").feature_extractor.conv_layers
combined = ConvLayer(wav2vec_layers[0].conv) * ConvLayer(wav2vec_layers[1].conv)
flatness = metrics.spectral_flatness(combined.weight).flatten()
plt.hist(flatness, bins=100)
```
